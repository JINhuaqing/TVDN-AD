```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
import os

%matplotlib inline
```

```python
os.chdir("/home/huaqingj/MyResearch/TVDN-AD/")
figPath = Path("./brainPlotFigs")
fils = sorted(list(figPath.glob("*.jpg")))
```

```python
_, axes = plt.subplots(3, 7, figsize=[60, 20])

fils = sorted(list(figPath.glob("CN*.jpg")))
for ix, fil in enumerate(fils):
    nam = "_".join(fil.stem.split("_")[1:])
    img = plt.imread(fil)
    axes[0, ix].imshow(img)
    axes[0, ix].axis("off")
    axes[0, ix].set_title(f"{nam}")
    if ix == 0:
        axes[0, ix].text(-0.1, 0.5, "Default", rotation=90,
                         horizontalalignment='center', verticalalignment='center', transform=axes[0, ix].transAxes)
    
fils = sorted(list(figPath.glob("*AD*.jpg")))
for ix, fil in enumerate(fils):
    img = plt.imread(fil)
    axes[1, ix].imshow(img)
    axes[1, ix].axis("off")
    if ix == 0:
        axes[1, ix].text(-0.1, 0.5, "AD", rotation=90,
                         horizontalalignment='center', verticalalignment='center', transform=axes[1, ix].transAxes)
    
fils = sorted(list(figPath.glob("*Ctrl*.jpg")))
for ix, fil in enumerate(fils):
    img = plt.imread(fil)
    axes[2, ix].imshow(img)
    axes[2, ix].axis("off")
    if ix == 0:
        axes[2, ix].text(-0.1, 0.5, "Control", rotation=90,
                         horizontalalignment='center', verticalalignment='center', transform=axes[2, ix].transAxes)
plt.savefig(f"./plots/mulBrains.jpg", bbox_inches='tight')
```
